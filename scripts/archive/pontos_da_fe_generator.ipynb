````python
import os
import yaml
import json
import numpy as np
import google.generativeai as genai
from PIL import Image, ImageDraw, ImageFilter
import scipy.io.wavfile as wav
from movielite import VideoWriter, ImageClip, TextClip, AudioClip, vfx, afx
import pictex

# CONFIGURAÇÃO - INSIRA SUA API KEY AQUI
GOOGLE_API_KEY = "SUA_API_KEY_AQUI"
genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel('gemini-1.5-pro')
````

### 🖼️ 1. Carregar Catálogo de Imagens
Lemos o JSON para passar a descrição dos ativos para a IA.

````python
with open("image_catalog.json", "r", encoding="utf-8") as f:
    catalog = json.load(f)

image_options = "\n".join([f"- {img['filename']}: {img['description']}" for img in catalog['images']])
print("Imagens carregadas no contexto:")
print(image_options)
````

### 🧠 2. Gerador de Roteiro Inteligente
O Gemini agora sabe exatamente quais imagens pode usar.

````python
def generate_script(tema):
    prompt = f"""
    Atue como Arquiteto de Conteúdo do canal 'Pontos da Fé'. 
    Crie um roteiro em formato YAML para o tema: {tema}

    DIRETRIZES VISUAIS (Use APENAS estes arquivos de imagem):
    {image_options}

    FORMATO YAML OBRIGATÓRIO:
    metadata:
      title: "Título"
      output_file: "video.mp4"
    segments:
      - type: "verse" | "narration" | "macro"
        text: "Texto do segmento"
        duration: 10
        start_time: 0
        bg_image: "nome_do_arquivo.jpeg"

    Retorne APENAS o YAML puro.
    """
    
    response = model.generate_content(prompt)
    yaml_text = response.text.replace('```yaml', '').replace('```', '').strip()
    
    with open("script_gerado.yaml", "w", encoding="utf-8") as f:
        f.write(yaml_text)
    
    return yaml_text
````
